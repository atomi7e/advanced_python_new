```python
names = ['Anna', 'Bob', 'Charlie', 'David', 'Eva']
scores = [85, 70, 95, 60, 82]

plt.plot(names, scores)
plt.xlabel('Имена студентов')
plt.ylabel('Баллы')
plt.title('Успеваемость (Линейный график)')

plt.show()
```

```python
plt.bar(names, scores, color='skyblue')

plt.xlabel('Имена студентов')
plt.ylabel('Баллы')
plt.title('Успеваемость (Столбчатая диаграмма)')

plt.show()
```
