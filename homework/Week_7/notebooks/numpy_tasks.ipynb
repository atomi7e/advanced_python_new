```python
import numpy as np
arr_1d = np.arange(1, 11)
print(arr_1d)
```

```python
arr_2d = np.random.randint(0, 10, (3, 3))
print("\nМассив 2D (3x3):")
print(arr_2d)
```

```python
print("Сумма всех элементов:", arr_2d.sum())
print("Среднее значение:", arr_2d.mean())
print("Максимальное значение:", arr_2d.max())
```

```python
arr_mult = arr_2d * 2
print("\nУмножили на 2:")
print(arr_mult)
```

```python
arr_replaced = arr_2d.copy()
arr_replaced[arr_replaced > 5] = 0

print("\nЗаменили числа > 5 на 0:")
print(arr_replaced)
```
