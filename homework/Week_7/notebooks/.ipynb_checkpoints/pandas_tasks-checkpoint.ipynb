```python
data = {
    'Name': ['Anna', 'Bob', 'Charlie', 'David', 'Eva'],
    'Age': [19, 22, 20, 21, 18],
    'Score': [85, 70, 95, 60, 82]
}

df = pd.DataFrame(data)
print("Наша таблица:")
print(df)
```

```python
print("Первые 3 строки:")
print(df.head(3))
```

```python
print("\nСтатистика:")
print(df.describe())
```

```python
avg_score = df['Score'].mean()
print("\nСредний балл:", avg_score)
```

```python
print("Студенты старше 20 лет:")
print(df[df['Age'] > 20])

print("\nСтуденты с баллом выше 80:")
print(df[df['Score'] > 80])
```
